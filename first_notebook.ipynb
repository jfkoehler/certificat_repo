### Collections aka Data Structures

- `list`: ordered and mutable
- `tuple`: ordered and immutable
- `dictionaries`: not ordered and mutable
- `sets`: not ordered and mutable

````python
#list
a_list = [1, 2, 3, 'lenny']
type(a_list)
````

````python
#tuple
a_tuple = (1, 2, 3, 'lenny')
type(a_tuple)
````

````python
#difference between tuple and list 
a_list[0] = 11
a_list
````

````python
a_tuple[0] = 11
````

````python
#dictionary = key, value pairs
a_dict = {'name': 'Lenny', 'age': 3, 'weight': 50}
type(a_dict)
````

````python
a_dict['name']
````

````python
#no order to a dictionary
a_dict[0]
````

````python
#set -- just the unique objects in a collection
a_set = {1, 1, 2, 3, 4, 4, 4, 'lenny'}
````

````python
a_set
````

````python
a_set[0]
````
